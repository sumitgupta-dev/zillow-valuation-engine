```python
# Let's pretend this is your heap after the loop is done.
# It contains 3 tuples: (Negative Distance, House Index)
my_heap = [(-40, 12), (-30, 4), (-20, 9)]

print("--- Step 1: What is in the heap? ---")
for item in my_heap:
    print(f"Tuple: {item} | Distance: {item[0]} | Index: {item[1]}")

print("\n--- Step 2: The List Comprehension ---")
# The 1-line shortcut
similar_indices = [idx for (neg_dist, idx) in my_heap]
print(f"Extracted Indices: {similar_indices}")

print("\n--- Step 3: What if we just used a normal loop? ---")
normal_indices = []
for item in my_heap:
    normal_indices.append(item[1]) # item[1] is the index
print(f"Normal Loop Indices: {normal_indices}")
```
